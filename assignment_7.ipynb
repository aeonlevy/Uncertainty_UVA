```python
rng = np.random.default_rng(100) 

def roll_6(): 
    sum = 0 
    on = 1 
    while on == 1: 
        roll = rng.integers(1,7) 
        sum += roll 
        if( roll < 6 ): 
            on = 0 
    return sum

sample = 1000000
results = [roll_6() for i in range(sample)]

print("Mean: ", np.mean(results))
print("Standard Deviation: ", np.std(results))
```

- Show, by example, that $v(\mathbb{E}[X]) \neq \mathbb{E}[v(X)]$, if $v(x) \neq a+bx$. For example, try $v(y) = y^2$ or $v(y)=\sqrt{y}$ with a Bernoulli or uniform or normally distributed random variable. This can be an important thing to remember: The expectation of a transformed random variable is not the transformation of the expected value.

```python
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 

p = 0.3
N = 100000 

X = np.random.binomial(1, p, size=N)

# v(x) = x^2
v = X**2

# v(E[X]) = (E[X])^2
v_E = (np.mean(X))**2

# E[v(X)] = E[X^2]
E_v = np.mean(v)

labels = ['E[X]^2', 'E[X^2]']
values = [v_E, E_v]

plt.figure(figsize=(10, 5))
bars = plt.bar(labels, values)

for bar, value in zip(bars, values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{value:.4f}",
        ha='center',
        va='bottom',
        fontsize=12
    )

plt.title("Comparison: v(E[X]) vs E[v(X)] for v(x)=x^2, X~Bernoulli(0.3)")
plt.ylabel("Value")
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()
```

```python
import scipy.stats as stats

mu_hat = np.mean(y)
sigma_hat = np.std(y, ddof=0) 


grid = np.linspace(9, 14, 100)


dist = stats.norm(loc=mu_hat, scale=sigma_hat)
pdf = dist.pdf(grid)
cdf = dist.cdf(grid)


fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.kdeplot(y, ax=axes[0], label="KDE")
sns.lineplot(x=grid, y=pdf, ax=axes[0], label="Normal PDF", linestyle="--")
axes[0].set_title("Density Comparison")
axes[0].legend()
sns.ecdfplot(y, ax=axes[1], label="ECDF")
sns.lineplot(x=grid, y=cdf, ax=axes[1], label="Normal CDF", linestyle="--")
axes[1].set_title("CDF Comparison")
axes[1].legend()

plt.tight_layout()
plt.show()
```

- Bootstrap the sampling density/distribution of your parameters.

```python
def mle(y):
    mu_hat = np.mean(y)
    sigma_hat = np.std(y, ddof=0)
    return mu_hat, sigma_hat


S = 5000
bootstrap_estimates = np.array([mle(y.sample(frac=1.0, replace=True)) 
                                for _ in range(S)])

mu_hat_samples = bootstrap_estimates[:, 0]
sigma_hat_samples = bootstrap_estimates[:, 1]


def plot_sampling_distribution(samples, label):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    sns.kdeplot(samples, ax=axes[0], label="KDE")
    axes[0].set_title(f"Sampling Density, ${label}$")
    axes[0].legend()

    sns.ecdfplot(samples, ax=axes[1], label="ECDF")
    axes[1].set_title(f"Sampling Distribution, ${label}$")
    axes[1].legend()

    plt.tight_layout()
    plt.show()

plot_sampling_distribution(mu_hat_samples, "\\hat{\\mu}")
plot_sampling_distribution(sigma_hat_samples, "\\hat{\\sigma}")
```
